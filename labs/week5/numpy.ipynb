```python
import matplotlib.pyplot as plt

x=np.arange(0,3*np.pi,0.1)
y_sin=np.sin(x)
y_cos=np.cos(x)


plt.plot(x,y_cos,label='cos')
plt.plot(x,y_sin,label='sin')

plt.xlabel('x axis label')
plt.ylabel('y axis label')
plt.title('sin and cos')

plt.show()
```
